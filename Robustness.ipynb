### Random swap : Swap two words randomly (excluding stop words) 

```python
!pip install nltk
```

```python
import nltk
nltk.download('stopwords')
```

```python
import random
from nltk.corpus import stopwords
lang="english"
def random_swap(sentence):
    words = sentence.split()
    stop_words = set(stopwords.words(lang))
    non_stopwords_indices = [i for i, word in enumerate(words) if word not in stop_words]
    if len(non_stopwords_indices) >= 3: # if we have list of 3 or more word 
        i, j = 0,1  # then select two word
        words[i], words[j] = words[j], words[i]  # Swap the words
    return " ".join(words)

files = os.listdir("data/data/crows2/")
for i in files:
    if "de_s" in i:
        lang="german"
    if "en_s" in i:
        lang="english"
    if "fr_s" in i:
        lang="french"
    if "nl_s" in i:
        lang="dutch"

    df = pd.read_csv("data/data/crows2/"+str(i))
    df["sent_more"] = df["sent_more"].apply(random_swap)
    df["sent_less"] = df["sent_less"].apply(random_swap)
    df.to_csv("data/data/crows/"+str(i))

```

#### Synonym Replacement- Replace common words with synonyms 

```python
import nltk
nltk.download('omw-1.4')
```

```python
import nltk
nltk.download('wordnet')
```

```python
import random
from nltk.corpus import stopwords,wordnet
lang="english"

def synonym_replacement(sentence):
  words = sentence.split()
  stop_words = set(stopwords.words(lang))
  for i, word in enumerate(words):
    if word not in stop_words:  # Exclude stop words
      synonyms = wordnet.synsets(word)
      if synonyms:
        synonym = synonyms[0].lemmas()[0].name()
        words[i] = synonym
  return " ".join(words)

files = os.listdir("data/data/crows2/")
for i in files:
    if "de_s" in i:
        lang="german"
    if "en_s" in i:
        lang="english"
    if "fr_s" in i:
        lang="french"
    if "nl_s" in i:
        lang="dutch"

    df = pd.read_csv("data/data/crows2/"+str(i))
    df["sent_more"] = df["sent_more"].apply(synonym_replacement)
    df["sent_less"] = df["sent_less"].apply(synonym_replacement)
    df.to_csv("data/data/crows/"+str(i))
```

 ### Noise Injection:

```python
import random
from nltk.corpus import stopwords,wordnet
lang="english"

def noise_injection(sentence):
  words = sentence.split()
  modified_words = []
  for word in words:
        random.seed(1)
        noise = ''.join("@+_+")
        modified_word = word + noise
        modified_words.append(modified_word)
        return " ".join(modified_words)

files = os.listdir("data/data/crows2/")
for i in files:
    if "de_s" in i:
        lang="german"
    if "en_s" in i:
        lang="english"
    if "fr_s" in i:
        lang="french"
    if "nl_s" in i:
        lang="dutch"

    df = pd.read_csv("data/data/crows2/"+str(i))
    df["sent_more"] = df["sent_more"].apply(noise_injection)
    df["sent_less"] = df["sent_less"].apply(noise_injection)
    df.to_csv("data/data/crows/"+str(i))
```

```python
def translate(sentence):
    translator = Translator()
    translation_fr = translator.translate(sentence, src='nl', dest='en').text
    translation_en_back = translator.translate(translation_fr, src='en', dest='nl').text
    return translation_en_back

files = os.listdir("data/data/crows2/")
for i in files:
    if "nl_s" in i:
        lang="english"
        df = pd.read_csv("data/data/crows2/"+str(i))
        df["sent_more"] = df["sent_more"].apply(translate)
        df["sent_less"] = df["sent_less"].apply(translate)
        df.to_csv("data/data/crows/"+str(i))
```
